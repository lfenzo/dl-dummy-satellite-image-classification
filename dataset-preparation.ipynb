```python
import os
import pandas as pd
from pathlib import Path

classes = [
    "cloudy",
    "desert",
    "green_area",
    "water",
]

dfs = []

for c in classes:
    directory = Path(os.path.join("data", c))
    files = [p.as_posix() for p in directory.iterdir() if p.is_file()]

    class_df = pd.DataFrame({"path": files})
    class_df["class"] = c

    dfs.append(class_df)

df = pd.concat(dfs, axis=0).reset_index(drop=True)
df
```

```python
df["class"].value_counts()
```

```python
from sklearn.model_selection import train_test_split

train_df, other = train_test_split(df, train_size=0.8, shuffle=True, stratify=df["class"])
valid_df, test_df = train_test_split(other, train_size=0.5)
```

## Sanity Check

```python
print(set(train_df.path).intersection(set(valid_df.path)))
print(set(train_df.path).intersection(set(test_df.path)))
```
